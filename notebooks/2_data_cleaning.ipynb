```python
import pandas as pd
import numpy as np

df = pd.read_csv('creditcard.csv')
print(f"Original shape: {df.shape}")
```

```python
before = len(df)
df_clean = df.drop_duplicates().reset_index(drop=True)
after = len(df_clean)

print(f"Rows before: {before}")
print(f"Rows after:  {after}")
print(f"Removed:     {before - after}")
```

```python
print("Class distribution after deduplication:")
print(df_clean['Class'].value_counts())
print(f"\nFraud percentage: {df_clean['Class'].mean() * 100:.4f}%")
```

```python
print("Missing values:", df_clean.isnull().sum().sum())
print("\nData types:")
print(df_clean.dtypes.value_counts())
```

```python
Q1 = df_clean['Amount'].quantile(0.25)
Q3 = df_clean['Amount'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df_clean[(df_clean['Amount'] < lower) | (df_clean['Amount'] > upper)]
print(f"IQR bounds: [{lower:.2f}, {upper:.2f}]")
print(f"Outliers detected in 'Amount': {len(outliers)} ({len(outliers)/len(df_clean)*100:.2f}%)")
print(f"Frauds among Amount outliers: {outliers['Class'].sum()}")
print("\nWe do not remove Amount outliers — many frauds are genuine outliers.")
```

```python
df_clean.to_csv('creditcard_clean.csv', index=False)

df_check = pd.read_csv('creditcard_clean.csv')
print(f"Reloaded shape: {df_check.shape}")
assert df_check.shape == df_clean.shape, "Shape mismatch!"
print("Shape verification passed.")
```
